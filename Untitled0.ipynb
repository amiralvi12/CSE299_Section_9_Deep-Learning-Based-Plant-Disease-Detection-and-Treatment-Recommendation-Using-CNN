```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dropout, Dense, BatchNormalization, GlobalAveragePooling2D, Input,Flatten, Activation
from tensorflow.keras.applications import VGG16
from tensorflow.keras.optimizers import Adam, SGD
from tensorflow.keras.activations import sigmoid, softmax, relu, tanh
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing import image
from sklearn.preprocessing import StandardScaler, LabelEncoder, MinMaxScaler
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
import os

print("Classes:", os.listdir(root_dir))
for cls in os.listdir(root_dir):
    print(cls, "→", len(os.listdir(os.path.join(root_dir, cls))))
```

```python

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_ds = datagen.flow_from_directory(
    root_dir,
    target_size=(224,224),
    class_mode='sparse',
    batch_size=128,   # changed from 64 → 128
    subset='training',
    shuffle=True
)

val_ds = datagen.flow_from_directory(
    root_dir,
    target_size=(224,224),
    batch_size=128,   # changed from 64 → 128
    class_mode='sparse',
    subset='validation',
    shuffle=False
)
```

```python
import os

print(os.listdir(root_dir))
```

```python
train_ds.class_indices
```

```python
# How many images did it find?
print("Train batches:", train_ds.samples)

# Class mapping (name -> id)
print("Classes:", train_ds.class_indices)

# Count images per class
from collections import Counter
counts = Counter(train_ds.classes)
id_to_name = {v:k for k,v in train_ds.class_indices.items()}
for cid, cnt in sorted(counts.items()):
    print(f"{id_to_name[cid]}: {cnt}")
```

```python
base = VGG16(weights='imagenet', include_top=False, input_shape=(224,224,3))
base.trainable = False
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, BatchNormalization, Dropout, LeakyReLU

vgg = Sequential([
    Input(shape=(224, 224, 3)),
    base,   # your VGG16 base
    Flatten(),
    Dense(128),                  # Dense without activation
    BatchNormalization(),
    LeakyReLU(alpha=0.1),        # LeakyReLU instead of ReLU
    Dropout(0.5),
    Dense(4, activation='softmax')  # output layer
])

vgg.summary()
```

```python
from tensorflow.keras.optimizers import AdamW

# Compile model with AdamW instead of Adam
vgg.compile(
    optimizer=AdamW(learning_rate=1e-4, weight_decay=1e-5),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
import tensorflow as tf
print("GPU Available:", tf.config.list_physical_devices('GPU'))
```

```python
history = vgg.fit(train_ds, validation_data=val_ds, epochs=40)
```

```python
plt.plot(history.history['accuracy'], color='red')
plt.plot(history.history['val_accuracy'], color='blue')
plt.show()
```

```python
plt.plot(history.history['loss'], color='red')
plt.plot(history.history['val_loss'], color='blue')
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import numpy as np

# Get predictions for validation set
y_pred = vgg.predict(val_ds)
y_pred_classes = np.argmax(y_pred, axis=1)

# True labels
y_true = val_ds.classes

# Confusion matrix
cm = confusion_matrix(y_true, y_pred_classes)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=list(val_ds.class_indices.keys()))
disp.plot(cmap=plt.cm.Blues, xticks_rotation=45)
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_true, y_pred_classes, target_names=list(val_ds.class_indices.keys())))
```
